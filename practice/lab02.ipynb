```python
from asyncio import Queue


q = Queue()
q = Queue()
q.push = q.put_nowait
q.pop = q.get_nowait
q.front = lambda: q._queue[0]
q.back = lambda: q._queue[-1]

q.push("34")
q.push("20")
q.pop()
q.push("69")
q.push("13")
q.front()
q.back()
q.push("20")
q.push("16")
q.pop()
q.push("69")
q.push("13")
q.front()
q.back()
```
